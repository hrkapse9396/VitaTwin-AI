```python
import numpy as np
from tensorflow.keras.models import load_model
```

```python
model = load_model("afib_cnn_model.h5")

X = np.load("X.npy")
y = np.load("y.npy")

print("Loaded Successfully")
```

```python
sample_index = 84

sample = X[sample_index].reshape(1,1250,1)
```

```python
probability = model.predict(sample)[0][0]

print("AFIB Probability:", probability)
```

```python
risk_score = probability * 100

if risk_score < 30:

    risk_level = "LOW RISK"

elif risk_score < 70:

    risk_level = "MEDIUM RISK"

else:

    risk_level = "HIGH RISK"

print("Risk Score:", round(risk_score,2))
print("Risk Level:", risk_level)
```

```python
if risk_level == "LOW RISK":

    recommendation = """
    Normal rhythm pattern detected.
    Continue routine monitoring.
    """

elif risk_level == "MEDIUM RISK":

    recommendation = """
    Mild irregular rhythm detected.
    Periodic cardiac monitoring is recommended.
    """

else:

    recommendation = """
    Significant AFIB characteristics detected.
    Detailed cardiac evaluation is recommended.
    """

print(recommendation)
```

```python
patient_summary = {
    "Patient_ID": "VT001",
    "ECG_Sample": sample_index,
    "AFIB_Probability": round(probability*100,2),
    "Risk_Score": round(risk_score,2),
    "Risk_Level": risk_level
}

patient_summary
```
